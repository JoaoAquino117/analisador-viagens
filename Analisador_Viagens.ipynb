```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('viagens_historico.csv')
df.head(3)
```

```python
total_gasto = df['Valor'].sum()

media_gasto = df['Valor'].mean()

print(f" O total gasto no período : R${total_gasto: .2f}")

print(f" Custo médio por viagem : R${media_gasto: .2f}")
```

```python
gasto_por_plataforma = df.groupby('App')['Valor'].sum().reset_index()

print("Gasto por Plataforma:")
print(gasto_por_plataforma)
```

```python
gastos_grafico = df.groupby('App')['Valor'].sum()

plt.figure(figsize = (8, 5))

gastos_grafico.plot(kind = 'bar', color = ['#00a2ed', 'black'])

plt.title('Gastos por Plataforma de Transporte', fontsize=14)
plt.xlabel('Plataforma', fontsize=12)
plt.ylabel('Custo Total (R$)', fontsize=12)
plt.xticks(rotation=0)
plt.show()
```
